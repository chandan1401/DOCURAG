# DocuRAG: Ingestion and Indexing Walkthrough

This notebook demonstrates how to load raw multimodal documents into the DocuRAG system, extract their content, and build the hybrid index.

```python
import os
from pathlib import Path
from docurag.pipeline import ingest_directory
from docurag.index.store import ChunkStore
from docurag.index.embedder import Embedder

# 1. Setup Environment
os.environ['GEMINI_API_KEY'] = 'YOUR_API_KEY_HERE'
RAW_DATA_DIR = 'data/raw/'

print(f'Starting ingestion from {RAW_DATA_DIR}...')
report = ingest_directory(RAW_DATA_DIR)
print('Ingestion Report:', report)
```

## Inspecting the Index
Now we load the store and check the stats to ensure we have a diverse set of multimodal chunks.

```python
embedder = Embedder()
store = ChunkStore(embedder)
store.load()

stats = store.stats()
import pandas as pd
df = pd.DataFrame.from_dict(stats, orient='index')
print(df)
```
