# DocuRAG: Retrieval and Grounded Answering

This notebook demonstrates how to use the DocuRAG pipeline to ask questions over a multimodal corpus.

```python
import os
from docurag.pipeline import ask

os.environ['GEMINI_API_KEY'] = 'YOUR_API_KEY_HERE'

questions = [
    "What is the main goal of DocuRAG?",
    "Can you find any tables regarding performance?",
    "What does the architectural diagram show?"
]

for q in questions:
    print(f'\nQuestion: {q}')
    result = ask(q)
    print(f'Answer: {result.answer}')
    print(f'Confidence: {result.confidence.label} ({result.confidence.score:.2f})')
    print(f'Sources: {[s.document for s in result.sources]}')
    print(f'Timing: {result.timing.total_ms}ms')
    print('-' * 40)
```
